```python
from Bio.SeqUtils.ProtParam import ProteinAnalysis

for protein in SeqIO.parse("tthermophilus_hb27.fasta.txt", "fasta"):
    seq = ProteinAnalysis(str(protein.seq))
    pi = seq.isoelectric_point()
    print(protein.id, pi)
```

```python
import pandas as pd

idlist = []
pilist = []
for protein in SeqIO.parse("tthermophilus_hb27.fasta.txt", "fasta"):
    seq = ProteinAnalysis(str(protein.seq))
    pi = seq.isoelectric_point()
    idlist.append(protein.id)
    pilist.append(pi)

dfpi = pd.DataFrame({"proteinid" : idlist, "pI": pilist})
print(dfpi)
print(len(dfpi))
```

```python
import pandas as pd

idlist = []
pilist = []
for protein in SeqIO.parse("tthermophilus_hb27.fasta.txt", "fasta"):
    seq = ProteinAnalysis(str(protein.seq))
    pi = seq.isoelectric_point()
    
    idlist.append(protein.id.split("|")[1])
    pilist.append(pi)

dfpi = pd.DataFrame({"proteinid" : idlist, "pI": pilist})
print(dfpi)
print(len(dfpi))
```
