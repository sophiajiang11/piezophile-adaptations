```python
from Bio import PDB
import pandas as pd
import os

parser = PDB.PDBParser(QUIET=True)
fracdisorder = {}
pdbfolder = os.path.expanduser('~/structures')
for pdbfile in os.listdir(pdbfolder):
    if not pdbfile.startswith('AF-') or not pdbfile.endswith('.pdb'):
        continue
    acc = pdbfile.split('-')[1]
    structure = parser.get_structure('', os.path.join(pdbfolder, pdbfile))
    bfactors = [a.get_bfactor() for a in structure.get_atoms() if a.get_name() == 'CA']
    fraction = sum(1 for b in bfactors if b < 50) / len(bfactors)
    fracdisorder[acc] = fraction

disorderdf = pd.DataFrame.from_dict(
    fracdisorder, orient='index', columns=['disorder_fraction'])
disorderdf.index.name = 'accession'
disorderdf.to_csv('disorderfractions.csv')
print(f'Done — {len(fracdisorder)} proteins')
```
